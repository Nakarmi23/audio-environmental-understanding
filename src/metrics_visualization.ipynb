```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
csv_path = "../runs/align/metrics.csv" 

df = pd.read_csv(csv_path)
df
```

```python
plt.figure()
plt.plot(df["epoch"], df["train_loss"], marker="o")
plt.xlabel("Epoch")
plt.ylabel("Training Loss")
plt.title("Training Loss over Epochs")
plt.grid(True)
plt.show()
```

```python
plt.figure()
plt.plot(df["epoch"], df["a2t_R@1"], marker="o", label="R@1")
plt.plot(df["epoch"], df["a2t_R@5"], marker="o", label="R@5")
plt.plot(df["epoch"], df["a2t_R@10"], marker="o", label="R@10")
plt.xlabel("Epoch")
plt.ylabel("Recall")
plt.title("Audio-to-Text Retrieval Performance")
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure()
plt.plot(df["epoch"], df["t2a_R@1"], marker="o", label="R@1")
plt.plot(df["epoch"], df["t2a_R@5"], marker="o", label="R@5")
plt.plot(df["epoch"], df["t2a_R@10"], marker="o", label="R@10")
plt.xlabel("Epoch")
plt.ylabel("Recall")
plt.title("Text-to-Audio Retrieval Performance")
plt.legend()
plt.grid(True)
plt.show()
```
